### 2.3 Exploratory & visual analyses

The play-by-play dataset appears to provide sufficient data to answer the team's questions. We focus on the 2019 season because it is the most recent in the dataset with complete documentation. High-level metrics about the play-by-play data for the 2019 season are shown below.


```python
# Data profile: shape, types, summary stats
pbp_df.info()
pbp_df = pbp_df.sort_values(["game_id", "game_seconds_remaining"], ascending=[True, False])

# For now, just use home-team win probability to determine game excitement
# We can make this more interesting, which will influence figure 1 (IMDB-scores)
# and Figure 4 (Momentum)
def game_excitement(group):
    wp = group["home_wp"].dropna().values
    if len(wp) < 2:
        return np.nan
    return np.abs(np.diff(wp)).sum()

excitement = (
    pbp_df.groupby("game_id").apply(game_excitement).rename("excitement").reset_index()
)
meta = games_df[["game_id", "home_team", "away_team", "home_score", "away_score", "week"]]
excitement = excitement.merge(meta, on="game_id")
excitement["matchup"] = excitement["away_team"] + " @ " + excitement["home_team"]
# excitement = excitement.sort_values("excitement", ascending=False)
```

```python
import matplotlib.colors as colors
from matplotlib.patches import Patch

# Figure 1: Game Rating
# Process the data
imdb_scores = {
    "Week 1": { "excitementScores": [], "gameScoreLabels": [] },
    "Week 2": { "excitementScores": [], "gameScoreLabels": [] },
    "Week 3": { "excitementScores": [], "gameScoreLabels": [] },
    "Week 4": { "excitementScores": [], "gameScoreLabels": [] },
    "Week 5": { "excitementScores": [], "gameScoreLabels": [] },
    "Week 6": { "excitementScores": [], "gameScoreLabels": [] },
    "Week 7": { "excitementScores": [], "gameScoreLabels": [] },
    "Week 8": { "excitementScores": [], "gameScoreLabels": [] },
    "Week 9": { "excitementScores": [], "gameScoreLabels": [] },
    "Week 10": { "excitementScores": [], "gameScoreLabels": [] },
    "Week 11": { "excitementScores": [], "gameScoreLabels": [] },
    "Week 12": { "excitementScores": [], "gameScoreLabels": [] },
    "Week 13": { "excitementScores": [], "gameScoreLabels": [] },
    "Week 14": { "excitementScores": [], "gameScoreLabels": [] },
    "Week 15": { "excitementScores": [], "gameScoreLabels": [] },
    "Week 16": { "excitementScores": [], "gameScoreLabels": [] },
    "Week 17": { "excitementScores": [], "gameScoreLabels": [] }
}
for index, game in excitement.iterrows():
    week = f"Week {game['week']}"
    imdb_scores[week]["excitementScores"].append(game['excitement'])
    gameScoreLabel = f"{game['away_team']} {game['away_score']} - {game['home_score']} {game['home_team']}"
    imdb_scores[week]["gameScoreLabels"].append(gameScoreLabel)

X_stage = imdb_scores.copy()
X = []
X_labels = []
for i in range(1, 18):
    numberOfGames = len(X_stage[f"Week {i}"]["excitementScores"])
    X_stage[f"Week {i}"]["excitementScores"].extend( [0] * (16 - numberOfGames) )
    X_stage[f"Week {i}"]["gameScoreLabels"].extend( [""] * (16 - numberOfGames) )
    X.append(X_stage[f"Week {i}"]["excitementScores"])
    X_labels.append(X_stage[f"Week {i}"]["gameScoreLabels"])

X = np.array(X).T
X_labels = np.array(X_labels).T

# Scale the data from 0-10
X_standardized = (X - np.mean(X)) / np.std(X)
X_scaled = (X_standardized - X_standardized.min()) / (X_standardized.max() - X_standardized.min()) * 10

# Create labels on each tile
X_labels_final = X_labels.astype(object).copy()
for i in range(X_labels_final.shape[0]):
    for j in range(X_labels_final.shape[1]):
        if X_labels_final[i, j] != "":
            X_labels_final[i, j] = f"{X_scaled[i, j]:.2f}\n{X_labels_final[i, j]}"

# Make the plot
bounds = [X_scaled.min(), 0.01, 1, 1.25, 1.75, 2.75, 3.2, 4, X_scaled.max()]

cmap = colors.ListedColormap([
    "white",
    "purple",
    "red",
    "orange",
    "yellow",
    "lightgreen",
    "darkgreen",
    "blue"
])

norm = colors.BoundaryNorm(bounds, cmap.N)

fig = plt.figure(figsize=(21,17))
ax = fig.add_subplot(111)
ax.imshow(X_scaled, cmap=cmap, norm=norm, interpolation='nearest')
numrows, numcols = X_scaled.shape

for row in range(numrows):
    for col in range(numcols):
        ax.text(
            col,
            row,
            f"{X_labels_final[row, col]}",
            ha="center",
            va="center",
            color="black",
            fontsize=6
        )

def format_coord(x, y):
    col = int(x+0.5)
    row = int(y+0.5)
    if col>=0 and col<numcols and row>=0 and row<numrows:
        z = X_scaled[row,col]
        return 'x=%1.4f, y=%1.4f, z=%1.4f'%(x, y, z)
    else:
        return 'x=%1.4f, y=%1.4f'%(x, y)

ax.format_coord = format_coord
# Draw borders around each tile
ax.set_xticks(np.arange(-0.5, numcols, 1), minor=True)
ax.set_yticks(np.arange(-0.5, numrows, 1), minor=True)

ax.grid(which="minor", color="black", linestyle="-", linewidth=1)

# Hide the minor tick marks themselves
ax.tick_params(which="minor", bottom=False, left=False)

# X-axis labels: 1 through 17
ax.set_xticks(np.arange(numcols))
ax.set_xticklabels(np.arange(1, numcols + 1))

# Hide the y-axis labels (and optionally the ticks)
ax.set_yticks([])

legend_elements = [
    Patch(facecolor="purple", edgecolor="black", label="Garbage"),
    Patch(facecolor="red",    edgecolor="black", label="Bad"),
    Patch(facecolor="orange", edgecolor="black", label="Regular"),
    Patch(facecolor="yellow", edgecolor="black", label="Good"),
    Patch(facecolor="lightgreen",  edgecolor="black", label="Great"),
    Patch(facecolor="darkgreen",   edgecolor="black", label="Awesome"),
    Patch(facecolor="blue",   edgecolor="black", label="Absolute Cinema")
]

ax.legend(
    handles=legend_elements,
    title="Legend",
    loc="upper left",          # or "center left", "lower right", etc.
    bbox_to_anchor=(1.02, 1),  # places it outside the plot
    borderaxespad=0
)
ax.set_xlabel("Week Number", fontsize=12)
ax.set_title("NFL Game Rating by Week", fontsize=14)
plt.show()
```

**Figure 1.** Colored tile chart showing game ratings throughout the 2019 NFL season.  Each tile displays the result of the game and an "excitement" score from 0-10.  A higher excitement score corresponds a higher swing in initial win probability.  This chart is inspired by the popular [_Series Graph_](https://seriesgraph.com/show/1399-game-of-thrones) charts of TV show episode scores on [_IMDb_](www.imdb.com).

This figure is gives a high level snapshot of an NFL season: What were the best games this season and when did they happen?  Games with high excitement scores are interesting to dive deeper into, such as investigating the momentum charts shown below and understanding where exactly the chaos from those games came from.


```python
# Figure 2: Success Metrics
plays = pbp_df[pbp_df["play_type"].isin(["run", "pass"])].copy()

team_game = plays.groupby(["game_id", "posteam"]).agg(
    epa_per_play=("epa", "mean"),
    success_rate=("epa", lambda x: (x > 0).mean()),
    n_plays=("epa", "size"),
    interceptions=("interception", "sum"),
    fumbles_lost=("fumble_lost", "sum"),
).reset_index()
team_game["turnovers"] = team_game["interceptions"] + team_game["fumbles_lost"]

third_downs = pbp_df[pbp_df["down"] == 3]
third_conv = (
    third_downs.groupby(["game_id", "posteam"])["third_down_converted"]
    .mean()
    .reset_index()
    .rename(columns={"third_down_converted": "third_down_rate"})
)
team_game = team_game.merge(third_conv, on=["game_id", "posteam"], how="left")

g = games_df[["game_id", "home_team", "away_team", "home_score", "away_score"]]
team_game = team_game.merge(g, on="game_id")
team_game["is_home"] = team_game["posteam"] == team_game["home_team"]
team_game["team_score"] = np.where(team_game["is_home"], team_game["home_score"], team_game["away_score"])
team_game["opp_score"] = np.where(team_game["is_home"], team_game["away_score"], team_game["home_score"])
team_game["win"] = (team_game["team_score"] > team_game["opp_score"]).astype(int)

metrics = ["epa_per_play", "success_rate", "turnovers", "third_down_rate"]
corrs = team_game[metrics + ["win"]].corr()["win"].drop("win").sort_values()

fig, ax = plt.subplots(1, 2, figsize=(13, 5))

colors = ["#c53030" if v < 0 else "#2f855a" for v in corrs.values]
ax[0].barh(corrs.index, corrs.values, color=colors)
ax[0].axvline(0, color="black", linewidth=0.8)
ax[0].set_title("Correlation with winning a game\n(team-game level, 2019)")
ax[0].set_xlabel("Pearson correlation with win (0/1)")

data_to_plot = [team_game.loc[team_game["win"] == 0, "epa_per_play"].dropna(),
                team_game.loc[team_game["win"] == 1, "epa_per_play"].dropna()]
bp = ax[1].boxplot(data_to_plot, labels=["Losses", "Wins"], patch_artist=True)
for patch, color in zip(bp["boxes"], ["#c53030", "#2f855a"]):
    patch.set_facecolor(color)
    patch.set_alpha(0.6)
ax[1].axhline(0, color="black", linewidth=0.8, linestyle="--")
ax[1].set_title("EPA/play: winning teams vs losing teams")
ax[1].set_ylabel("Offensive EPA per play")

plt.tight_layout()
plt.show()

print("\nCorrelation with win:")
print(corrs.to_string())
```

**Figure 2.** Overview of metrics with strong correlation to team success

The data shows that Expected Points Added (EPA) per play is the strongest indicator of team success (winning), while turnovers are strongly correlated to losing. These correlations make sense, as the team with more points at the end of regulation is the winner in the NFL. Teams who struggle to score, or give their opponents more oppourtunities to score are more likely to lose.


```python
# Figure 3: Game location and rest
game_dates = pbp_df.groupby("game_id")["game_date"].first().reset_index()
game_dates["game_date"] = pd.to_datetime(game_dates["game_date"])
gd = games_df.merge(game_dates, on="game_id")
gd["home_win"] = (gd["home_score"] > gd["away_score"]).astype(int)

home = gd[["game_id", "week", "game_date", "home_team", "home_score", "away_score"]].rename(
    columns={"home_team": "team", "home_score": "team_score", "away_score": "opp_score"})
home["is_home"] = True
away = gd[["game_id", "week", "game_date", "away_team", "away_score", "home_score"]].rename(
    columns={"away_team": "team", "away_score": "team_score", "home_score": "opp_score"})
away["is_home"] = False
long = pd.concat([home, away], ignore_index=True)
long["win"] = (long["team_score"] > long["opp_score"]).astype(int)
long = long.sort_values(["team", "game_date"])
long["rest_days"] = long.groupby("team")["game_date"].diff().dt.days

bucket = pd.cut(
    long["rest_days"], bins=[0, 6, 7, 9, 20],
    labels=["<=6\n(short week)", "7\n(normal)", "8-9\n(mini extra rest)", "10+\n(bye week)"]
)
rest_win = long.groupby(bucket, observed=True)["win"].agg(["mean", "count"])

fig, ax = plt.subplots(1, 2, figsize=(13, 5))

home_away_rate = pd.Series({"Home": gd["home_win"].mean(), "Away": 1 - gd["home_win"].mean()})
ax[0].bar(home_away_rate.index, home_away_rate.values, color=["#2b6cb0", "#a0aec0"])
ax[0].axhline(0.5, color="black", linestyle="--", linewidth=0.8)
ax[0].set_ylim(0, 1)
ax[0].set_title("Home field advantage, 2019\n(win rate)")
ax[0].set_ylabel("Win rate")
for i, v in enumerate(home_away_rate.values):
    ax[0].text(i, v + 0.02, f"{v:.1%}", ha="center")

ax[1].bar(rest_win.index.astype(str), rest_win["mean"], color="#805ad5")
ax[1].axhline(0.5, color="black", linestyle="--", linewidth=0.8)
ax[1].set_ylim(0, 1)
ax[1].set_title("Win rate by days of rest\nbefore the game")
ax[1].set_ylabel("Win rate")
for i, (m, c) in enumerate(zip(rest_win["mean"], rest_win["count"])):
    ax[1].text(i, m + 0.02, f"n={c}", ha="center", fontsize=8)

plt.tight_layout()
plt.show()

print("\nHome win rate:", gd["home_win"].mean())
print("\nWin rate by rest bucket:")
print(rest_win.to_string())
```

**Figure 3.** Overview of location and rest days effect on team success

The data shows that home teams had a slight advantage in 2019 though arguably the result in not signficant. It is commonly accepted that home field leads to higher win probability, the numbers don't support a massive correlation for the 2019 season.

It is also generally accepted that teams with more rest between games have more time to prepare and to recover from games or injuries, however the data shows that win rate was inversely correlated to rest days in this dataset.

```python
# Figure 4: Momentum
def plot_momentum(game_id):
  game_row = excitement.loc[excitement['game_id'] == game_id].iloc[0]
  g_plays = pbp_df[pbp_df["game_id"] == game_id].sort_values("game_seconds_remaining", ascending=False)
  g_plays = g_plays.dropna(subset=["home_wp"])

  elapsed_min = (3600 - g_plays["game_seconds_remaining"]) / 60.0

  fig, ax = plt.subplots(figsize=(12, 5))
  ax.plot(elapsed_min, g_plays["home_wp"], color="#2b6cb0", linewidth=1.6)
  ax.fill_between(elapsed_min, g_plays["home_wp"], 0.5,
                  where=(g_plays["home_wp"] >= 0.5), color="#2b6cb0", alpha=0.25)
  ax.fill_between(elapsed_min, g_plays["home_wp"], 0.5,
                  where=(g_plays["home_wp"] < 0.5), color="#c53030", alpha=0.25)
  ax.axhline(0.5, color="black", linewidth=0.8, linestyle="--")
  for q in [15, 30, 45, 60]:
      ax.axvline(q, color="gray", linewidth=0.6, linestyle=":")
  ax.set_ylim(0, 1)
  ax.set_xlim(0, elapsed_min.max())
  ax.set_xlabel("Game minute elapsed")
  ax.set_ylabel(f"{game_row['home_team']} win probability")
  ax.set_title(
      f"Momentum chart: {game_row['away_team']} @ {game_row['home_team']} "
      f"(wk {game_row['week']}, final {game_row['home_score']}-{game_row['away_score']})\n"
      f"GEI={game_row['excitement']:.1f}"
  )
  plt.tight_layout()
  plt.show()

  print(f"\nMomentum chart built for game_id={game_id}: "
        f"{game_row['away_team']} @ {game_row['home_team']} (wk {game_row['week']})")

plot_momentum(excitement.iloc[0]["game_id"])
plot_momentum(excitement.iloc[-1]["game_id"])

```

**Figure 4.** Game momentum graphs

These graphs show the momentum swings in individual NFL games in terms of the win probability for the home team. Large, immediate changes in win probability, especially near the end of a game, is likely to correlate heavily with the game's excitement index in the final analysis.

### 3.3 Limitations and how the analysis could be improved

#### Limitations

Both the FIFA World Cup and NFL datasets provide valuable opportunities for sports analytics, but each has inherent limitations. The FIFA World Cup dataset is well organized and contains detailed information on teams, players, matches, and match events. However, because it represents a single international tournament, the number of observations is relatively small, which limits the statistical power of exploratory analyses and predictive modeling.

The NFL play-by-play dataset offers a much larger volume of data but is not without limitations. While it captures detailed in-game events, it does not account for every factor that may influence game outcomes, such as player injuries, weather conditions, coaching strategies, or team chemistry. Additionally, because the data reflects historical games, models developed from it may not fully generalize to future seasons as team rosters, strategies, and league rules evolve. Like many publicly available datasets, it may also contain missing values, inconsistencies, or biases introduced during the data collection process.

#### Improvements

The analysis could be strengthened by expanding the available data and incorporating additional contextual information. For the FIFA dataset, combining data from multiple World Cup tournaments or including international league and qualifying matches would provide a larger sample for more reliable statistical analysis. For the NFL dataset, integrating complementary data sources such as injury reports, weather conditions, betting odds, player tracking data, or team rankings could improve the accuracy and interpretability of predictive models.

Future work could also explore more advanced analytical techniques, including feature engineering, ensemble machine learning methods, clustering, and time-series analysis to better capture trends in team and player performance. Model performance should be validated using cross-validation and testing on independent seasons or tournaments to improve reliability. Finally, the findings could be incorporated into an interactive dashboard that allows users to explore statistics, visualize trends, and generate predictions, making the analysis more accessible to coaches, analysts, researchers, media organizations, and sports fans.

The NFL play-by-play dataset was selected because it provides a substantially larger sample size and a more comprehensive record of in-game events than the FIFA World Cup dataset. Since the project aims to identify meaningful relationships and potentially develop predictive models, having data from thousands of games across multiple seasons allows for more reliable statistical analyses and better model performance.

The play-by-play data also supports a wider range of analytical techniques, including exploratory data analysis, feature engineering, predictive modeling, clustering, and trend analysis. Its level of detail enables the team to examine how specific game situations such as field position, turnovers, penalties, or offensive efficiency, contribute to game outcomes.

While the FIFA dataset remains a valuable resource for tournament analysis and sports visualization, the NFL dataset better aligns with the project's objective of conducting an in-depth analysis supported by a large, information-rich dataset capable of producing statistically meaningful results and supporting the development of a predictive sports analytics application.
